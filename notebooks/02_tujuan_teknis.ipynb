**Kode unit:** J.62DMI00.002.1 · SKKNI No. 299 Tahun 2020 · Proyek: *Prediksi Arah Harian IHSG*

**Tujuan unit ini:** menerjemahkan tujuan bisnis menjadi pekerjaan data science yang bisa diukur.

## Jenis masalah
**Klasifikasi biner** (dua kelas):

| Kelas | Arti |
|---|---|
| 1 = **Naik** | IHSG besok ditutup lebih tinggi daripada hari ini |
| 0 = **Turun** | IHSG besok ditutup sama atau lebih rendah |

## Data yang dipakai sebagai input (fitur)
Perubahan harian (%) dari 6 variabel pasar pada hari ini: **IHSG, S&P 500, VIX, minyak, emas, dan kurs USD/IDR**.
Semua data ini sudah tersedia sebelum BEI dibuka keesokan paginya. Setelah review iterasi pertama, kurs dikeluarkan
sehingga model akhir memakai 5 fitur (lihat Unit 9–11).

## Ukuran keberhasilan (ditetapkan sebelum membuat model)
| Ukuran | Target | Penjelasan sederhana |
|---|---|---|
| Accuracy | ≥ 0,55 | Persentase tebakan yang benar |
| F1-macro | ≥ 0,55 | Ketepatan rata-rata untuk kelas Naik dan Turun |
| ROC-AUC | ≥ 0,58 | Kemampuan membedakan hari naik dan turun (0,5 = asal tebak) |
| Baseline | Lebih tinggi | Accuracy harus mengalahkan tebakan "selalu naik" |

Target sengaja tidak tinggi karena arah pasar saham harian memang sulit ditebak.

## Cara menguji model
- Data diurutkan menurut waktu: **80% awal untuk melatih** model, **20% terakhir untuk menguji**.
- Data **tidak diacak**, karena kita tidak boleh memakai data masa depan untuk menebak masa lalu.
