`target = 1` jika IHSG pada hari bursa berikutnya ditutup **lebih tinggi** daripada hari ini, selain itu `0`.

## Contoh pembentukan target

```python
df = pd.read_csv('../data/raw/market_data.csv', parse_dates=['Date'])
contoh = df[['Date', 'IHSG']].dropna()
contoh = contoh[contoh['IHSG'].diff() != 0].head(7).copy()
contoh['IHSG_besok'] = contoh['IHSG'].shift(-1)
contoh['target'] = (contoh['IHSG_besok'] > contoh['IHSG']).astype(int)
contoh.head(6)
```

## Kesimpulan
Objek data adalah **hari bursa BEI** dengan **5 variabel pasar harian**. Yang ditebak adalah **arah IHSG hari bursa
berikutnya** (1 = naik, 0 = turun).
