## Cara memilih model terbaik
Model dibandingkan dengan **validasi silang deret waktu (`TimeSeriesSplit`, 5 bagian)** pada data latih:
model dilatih dengan data lama lalu diuji pada periode sesudahnya, diulang 5 kali. Model dengan **accuracy rata-rata
tertinggi** dipilih. Data uji belum dipakai sama sekali.

```python
from sklearn.model_selection import TimeSeriesSplit

latih = pd.read_csv('../data/processed/latih.csv', parse_dates=['Date'])
cv = TimeSeriesSplit(n_splits=5)
for i, (idx_latih, idx_val) in enumerate(cv.split(latih), 1):
    print(f"Bagian {i}: latih {latih['Date'].iloc[idx_latih[0]].year}–{latih['Date'].iloc[idx_latih[-1]].year} "
          f"→ validasi {latih['Date'].iloc[idx_val[0]].year}–{latih['Date'].iloc[idx_val[-1]].year}")
```
