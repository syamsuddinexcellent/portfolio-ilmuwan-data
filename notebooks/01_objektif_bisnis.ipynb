# Unit 1: Menentukan Objektif Bisnis
**Kode unit:** J.62DMI00.001.1 · SKKNI No. 299 Tahun 2020 · Proyek: *Prediksi Arah Harian IHSG*

**Tujuan unit ini:** menjelaskan masalah bisnis, siapa penggunanya, dan kapan proyek dianggap berhasil.

## Latar belakang
IHSG adalah indeks utama Bursa Efek Indonesia (BEI). Pergerakannya dipengaruhi pasar global, misalnya bursa saham
Amerika (S&P 500), indeks ketakutan pasar global (VIX), harga minyak, dan harga emas.

BEI tutup pukul **16.00 WIB**, sedangkan bursa Amerika baru buka malam harinya. Jadi, setiap pagi sebelum BEI dibuka,
kita sudah tahu apa yang terjadi di pasar global semalam.

## Pertanyaan bisnis
> **Dengan data pasar yang sudah tersedia pagi hari, bisakah kita memperkirakan apakah IHSG akan ditutup naik hari itu?**

## Pengguna hasil proyek
| Pengguna | Kebutuhan |
|---|---|
| Analis riset sekuritas | Bahan laporan pagi (*morning brief*) |
| Manajer investasi | Gambaran arah pasar sebelum mengambil keputusan |
| Manajemen risiko | Peringatan dini saat pasar berpeluang turun |

## Tujuan bisnis
1. Memberi perkiraan arah IHSG (naik atau turun) sebelum BEI dibuka.
2. Mengetahui faktor pasar mana yang paling berpengaruh.

## Kriteria keberhasilan
- Perkiraan model **lebih baik daripada tebakan sederhana** (misalnya selalu menebak "naik").
- Hasil dapat dijelaskan kepada pengguna yang bukan ahli data.

## Risiko
| Risiko | Cara mengatasi |
|---|---|
| Pasar saham sulit ditebak | Target dibuat realistis dan dibandingkan dengan tebakan sederhana |
| Kualitas data kurang baik | Data diperiksa (Unit 4) dan dibersihkan (Unit 6) |
| Hasil disalahartikan sebagai saran investasi | Hasil diposisikan sebagai alat bantu, bukan saran investasi |
